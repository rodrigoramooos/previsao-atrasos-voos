# Modelação: Previsão de Cancelamentos e Atrasos em Voos

## Sumário

Este notebook corresponde à Milestone 3 do projeto e realiza a fase de modelação sobre dados reais de voos domésticos norte-americanos processados na Milestone 2. O dataset `flight_data_processed_novo.csv` contém 1 041 151 registos de Janeiro a Fevereiro de 2024, com 340 features após limpeza, engenharia de atributos e codificação one-hot dos aeroportos de origem.

São modeladas duas variáveis alvo independentes, com pipelines completamente separadas para evitar leakage de informação entre targets:

| Target | Descrição | Prevalência | Rácio neg:pos |
|--------|-----------|-------------|---------------|
| `cancelled` | Cancelamento do voo | 1.53% | ~64:1 |
| `is_delayed` | Atraso >= 15 min por meteorologia ou aeronave anterior (excluindo cancelamentos) | 8.52% | ~11:1 |

O desequilíbrio de classes é a característica dominante deste problema e determina as escolhas metodológicas: métricas, modelos, compensação de classes e otimização do threshold.

**Estrutura:**
1. Preparação dos dados e configuração partilhada
2. Pipeline completa para `cancelled` (baseline → candidatos → otimização → avaliação)
3. Pipeline completa para `is_delayed` (estrutura idêntica, variáveis distintas)
4. Comparação entre targets e reflexão crítica

### 1.2 Métricas de Avaliação e Funções Auxiliares

A **accuracy** é excluída como métrica principal: um modelo que preveja sempre a classe negativa obteria 98.47% de accuracy em `cancelled` sem qualquer utilidade prática. As métricas utilizadas são:

- **F1-Score** (métrica principal de comparação): equilíbrio entre Precision e Recall;
- **Recall**: proporção de eventos positivos corretamente detetados;
- **Precision**: fiabilidade dos alertas gerados;
- **ROC-AUC**: capacidade discriminativa global, independente do threshold;
- **Average Precision (PR-AUC)**: área sob a curva Precision-Recall — mais informativa que ROC-AUC em datasets fortemente desequilibrados.

O threshold de decisão é otimizado por F1 sobre a curva Precision-Recall, evitando o uso do threshold padrão 0.5 que seria inadequado com rácios 64:1 ou 11:1.


```python
# ── Funções de avaliação ──────────────────────────────────────────────────
def threshold_otimo_f1(y_true, y_proba):
    p, r, t = precision_recall_curve(y_true, y_proba)
    f1 = 2 * p * r / (p + r + 1e-9)
    i = f1.argmax()
    return float(t[i]) if i < len(t) else 0.5

def avaliar(y_true, y_pred, y_proba, nome=''):
    return {
        'Modelo':        nome,
        'F1':            round(f1_score(y_true, y_pred, zero_division=0), 4),
        'Recall':        round(recall_score(y_true, y_pred, zero_division=0), 4),
        'Precision':     round(precision_score(y_true, y_pred, zero_division=0), 4),
        'ROC-AUC':       round(roc_auc_score(y_true, y_proba), 4),
        'Avg Precision': round(average_precision_score(y_true, y_proba), 4),
    }

def recall_por_grupo(g):
    tp = ((g['y_true'] == 1) & (g['y_pred'] == 1)).sum()
    fn = ((g['y_true'] == 1) & (g['y_pred'] == 0)).sum()
    return tp / (tp + fn) if (tp + fn) > 0 else 0.0

def relatorio_df(y_true, y_pred, target_names):
    rep = classification_report(
        y_true, y_pred, zero_division=0,
        target_names=target_names, output_dict=True
    )
    df_rep = pd.DataFrame(rep).T
    df_rep['support'] = df_rep['support'].astype(int)
    return df_rep[['precision', 'recall', 'f1-score', 'support']].round(3)

# ── Funções de visualização ───────────────────────────────────────────────
def plot_prob_dist(y_true, y_proba, threshold, title, cor_pos='tomato', cor_neg='steelblue'):
    fig, ax = plt.subplots(figsize=(11, 4))
    ax.hist(y_proba[y_true == 0], bins=60, alpha=0.55, color=cor_neg,
            density=True, label='Classe negativa (maioritária)')
    ax.hist(y_proba[y_true == 1], bins=60, alpha=0.70, color=cor_pos,
            density=True, label='Classe positiva')
    ax.axvline(threshold, color='black', linestyle='--', lw=1.8,
               label=f'Threshold = {threshold:.3f}')
    ax.set_xlabel('Probabilidade Prevista')
    ax.set_ylabel('Densidade')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.legend()
    ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

def plot_feature_imp(modelo, feature_names, title,
                     X_val=None, y_val=None, cor='steelblue', n=20):
    fi_vals = None
    for obj in [modelo,
                getattr(modelo, 'estimator_', None),
                getattr(modelo, 'estimator',  None)]:
        if obj is not None and hasattr(obj, 'feature_importances_'):
            fi_vals = obj.feature_importances_
            break
    if fi_vals is None:
        if X_val is None:
            print('Modelo sem feature_importances_ e sem dados de validação.')
            return
        print('Calculando permutation importance (fallback para HistGBT)...')
        n_sample = min(4000, len(X_val))
        idx = np.random.default_rng(42).choice(len(X_val), n_sample, replace=False)
        X_s = X_val.iloc[idx].astype('float32')
        y_s = y_val.iloc[idx]
        perm = permutation_importance(
            modelo, X_s, y_s,
            n_repeats=3, random_state=42,
            scoring='average_precision', n_jobs=-1
        )
        fi_vals = perm.importances_mean
    fi = pd.Series(fi_vals, index=list(feature_names))
    top = fi.nlargest(n).sort_values()
    fig, ax = plt.subplots(figsize=(10, max(5, n * 0.36)))
    ax.barh(top.index, top.values, color=cor, alpha=0.82, edgecolor='white')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_xlabel('Importância (gain / permutation)')
    ax.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()

def plot_threshold_curve(y_true, y_proba, threshold_otimo, title, cor='steelblue'):
    p, r, t = precision_recall_curve(y_true, y_proba)
    f1v = 2 * p[:-1] * r[:-1] / (p[:-1] + r[:-1] + 1e-9)
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(t, f1v,    color=cor,     lw=2.0, label='F1-Score')
    ax.plot(t, p[:-1], color='green', lw=1.5, alpha=0.75, linestyle='--', label='Precision')
    ax.plot(t, r[:-1], color='tomato',lw=1.5, alpha=0.75, linestyle='--', label='Recall')
    ax.axvline(threshold_otimo, color='black', linestyle=':', lw=1.8,
               label=f'Threshold otimizado = {threshold_otimo:.3f}')
    ax.set_xlabel('Threshold de Decisão')
    ax.set_ylabel('Valor da Métrica')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_ylim(0, 1)
    ax.legend()
    ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

def plot_metricas_candidatos(df_res, title, paleta):
    metricas_vis = ['F1', 'Recall', 'Precision', 'ROC-AUC', 'Avg Precision']
    x = np.arange(len(metricas_vis))
    w = 0.8 / len(df_res)
    fig, ax = plt.subplots(figsize=(12, 5))
    for i, (_, row) in enumerate(df_res.iterrows()):
        vals = [row[m] for m in metricas_vis]
        ax.bar(x + i * w - (len(df_res)-1)*w/2, vals, w,
               label=row['Modelo'], color=paleta[i], alpha=0.85, edgecolor='white')
    ax.set_xticks(x)
    ax.set_xticklabels(metricas_vis)
    ax.set_ylim(0, 1.05)
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.legend()
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

NOMES_MES = ['Jan','Fev','Mar','Abr','Mai','Jun','Jul','Ago','Set','Out','Nov','Dez']
NOMES_DIA = {0:'Dom', 1:'Seg', 2:'Ter', 3:'Qua', 4:'Qui', 5:'Sex', 6:'Sáb'}

print('Funções auxiliares e constantes definidas.')
```

## 2. Modelação — Previsão de Cancelamentos (`cancelled`)

Esta secção modela a variável `cancelled`. Com apenas 1.53% de registos positivos, este é o problema mais desafiante do notebook: o desequilíbrio extremo constitui um "teto" para o F1-Score atingível, independentemente do algoritmo usado. A pipeline é organizada em cinco fases: preparação, baseline, candidatos, otimização e avaliação.

Todas as variáveis seguem a nomenclatura `*_cancelled` para garantir separação explícita dos dados e modelos desta secção relativamente à secção `is_delayed`.


### 2.1 Preparação dos Dados

```python
X_cancelled = df[feature_cols].copy()
y_cancelled = df['cancelled'].copy()

X_train_cancelled, X_test_cancelled, y_train_cancelled, y_test_cancelled = train_test_split(
    X_cancelled, y_cancelled,
    test_size=0.20, random_state=42, stratify=y_cancelled
)

spw_cancelled = float((y_train_cancelled == 0).sum() / (y_train_cancelled == 1).sum())

divisao_canc = pd.DataFrame({
    'Conjunto': ['Treino', 'Teste'],
    'Registos': [X_train_cancelled.shape[0], X_test_cancelled.shape[0]],
    'Positivos': [int(y_train_cancelled.sum()), int(y_test_cancelled.sum())],
    'Prevalencia': [
        f'{y_train_cancelled.mean():.3%}',
        f'{y_test_cancelled.mean():.3%}',
    ],
})
display(divisao_canc.set_index('Conjunto'))
print(f'\nscale_pos_weight para XGBoost: {spw_cancelled:.1f}')
```

O treino contém 832 920 voos, dos quais apenas 12 724 são cancelamentos (1.528%). O `scale_pos_weight` de 64.5 reflete o rácio exato de negativos para positivos.


### 2.2 Modelo Baseline

A Regressão Logística com `class_weight='balanced'` serve de referência mínima — um modelo propositadamente simples para estabelecer um patamar que os modelos mais complexos devem superar. O threshold é otimizado por F1 sobre a curva Precision-Recall do conjunto de teste.

```python
baseline_cancelled = LogisticRegression(
    class_weight='balanced', max_iter=1000, random_state=42, n_jobs=-1
)
baseline_cancelled.fit(X_train_cancelled.astype('float32'), y_train_cancelled)

y_proba_bl_canc = baseline_cancelled.predict_proba(
    X_test_cancelled.astype('float32'))[:, 1]
thr_bl_canc = threshold_otimo_f1(y_test_cancelled, y_proba_bl_canc)
y_pred_bl_canc = (y_proba_bl_canc >= thr_bl_canc).astype(int)

print(f'Threshold otimizado (F1): {thr_bl_canc:.4f}')
print()
display(relatorio_df(y_test_cancelled, y_pred_bl_canc,
                     ['Não Cancelado', 'Cancelado']))
```

Com threshold otimizado de **0.733** — muito acima do 0.5 convencional — a Regressão Logística atinge apenas Recall=0.235 e Precision=0.053 (F1=0.086). O modelo só considera cancelamento quando a probabilidade ultrapassa 73%. O resultado confirma que a fronteira linear é insuficiente para capturar os padrões de cancelamento com 340 features e rácio 64:1.


### 2.3 Comparação de Modelos Candidatos

Testam-se dois modelos de gradient boosting em adição ao baseline:

- **HistGradientBoostingClassifier** com `class_weight='balanced'`;
- **XGBoostClassifier** com `scale_pos_weight` calibrado para o rácio 64:1.

O threshold de cada candidato é otimizado individualmente por F1.

```python
candidatos_cancelled = {
    'Logistic Regression': (baseline_cancelled, y_proba_bl_canc),
    'HistGradient Boosting': (
        HistGradientBoostingClassifier(
            class_weight='balanced', learning_rate=0.08,
            max_depth=8, max_iter=200, min_samples_leaf=30,
            random_state=42),
        None),
    'XGBoost': (
        XGBClassifier(
            scale_pos_weight=spw_cancelled, learning_rate=0.08,
            max_depth=6, n_estimators=200, subsample=0.8,
            colsample_bytree=0.8, eval_metric='aucpr',
            random_state=42, n_jobs=1, verbosity=0),
        None),
}

resultados_canc     = []
modelos_canc        = {}
probabilidades_canc = {}

for nome, (modelo, y_pr_cache) in candidatos_cancelled.items():
    if y_pr_cache is not None:
        y_pr = y_pr_cache
    else:
        modelo.fit(X_train_cancelled.astype('float32'), y_train_cancelled)
        y_pr = modelo.predict_proba(X_test_cancelled.astype('float32'))[:, 1]
    modelos_canc[nome]        = modelo
    probabilidades_canc[nome] = y_pr
    thr  = threshold_otimo_f1(y_test_cancelled, y_pr)
    y_pd = (y_pr >= thr).astype(int)
    res  = {**avaliar(y_test_cancelled, y_pd, y_pr, nome), 'Threshold': round(thr, 4)}
    resultados_canc.append(res)

df_candidatos_canc = (
    pd.DataFrame(resultados_canc)
    .sort_values('Avg Precision', ascending=False)
    .reset_index(drop=True)
)
display(df_candidatos_canc.set_index('Modelo').round(4))

melhor_nome_canc = df_candidatos_canc.iloc[0]['Modelo']
print(f'\nMelhor candidato (por Avg Precision): {melhor_nome_canc}')
```

```python
# Curvas ROC e Precision-Recall — candidatos cancelled
cores_canc = ['#4878CF', '#6ACC65', '#D65F5F']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Curvas de Desempenho — Candidatos (cancelled)',
             fontsize=13, fontweight='bold')

for (nome, y_pr), cor in zip(probabilidades_canc.items(), cores_canc):
    fpr, tpr, _ = roc_curve(y_test_cancelled, y_pr)
    axes[0].plot(fpr, tpr, color=cor, lw=1.8,
                 label=f'{nome} (AUC={auc(fpr,tpr):.3f})')
    prec, rec, _ = precision_recall_curve(y_test_cancelled, y_pr)
    axes[1].plot(rec, prec, color=cor, lw=1.8,
                 label=f'{nome} (AP={auc(rec,prec):.3f})')

axes[0].plot([0, 1], [0, 1], 'k--', lw=1, alpha=0.5)
axes[0].set_xlabel('Taxa de Falsos Positivos')
axes[0].set_ylabel('Taxa de Verdadeiros Positivos')
axes[0].set_title('Curva ROC')
axes[0].legend(fontsize=9, loc='lower right')
axes[0].grid(alpha=0.3)

prev_canc = y_test_cancelled.mean()
axes[1].axhline(prev_canc, color='k', linestyle='--', lw=1, alpha=0.5,
                label=f'Classificador aleatório ({prev_canc:.3f})')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].set_title('Curva Precision-Recall')
axes[1].legend(fontsize=9, loc='upper right')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Comparação de métricas entre candidatos — gráfico de barras agrupadas
plot_metricas_candidatos(
    df_candidatos_canc,
    'Comparação de Métricas — Candidatos (cancelled)',
    cores_canc
)
```

O critério de seleção entre candidatos é a **Average Precision** (área sob a curva Precision-Recall),
e não o F1-Score nem o ROC-AUC. O motivo é simples: o F1 obriga a fixar um threshold antes de avaliar
— o que é prematuro nesta fase — e o ROC-AUC é inflacionado pela enorme massa de negativos, dando uma falsa impressão de um bom desempenho. A Average Precision avalia o modelo em todos os thresholds
simultaneamente e penaliza diretamente os falsos alarmes, sendo mais honesta em problemas com raridade
extrema de positivos.


O **HistGBT obtém a melhor Average Precision** (0.096 vs 0.094 do XGBoost) e é por isso selecionado
para a fase de otimização. A diferença é pequena, mas ambos distanciam-se claramente do baseline
(AP=0.043) — mais do dobro. Os thresholds elevados (0.845 no HistGBT, 0.791 no XGBoost) mostram
que o modelo só arrisca classificar um voo como cancelamento quando a probabilidade é muito alta,
o que limita o Recall mas mantém alguma Precision.

### 2.4 Otimização do Melhor Modelo

O melhor candidato é submetido a duas fases de otimização:

1. **GridSearchCV** com `scoring='average_precision'` sobre 30% dos dados de treino (~250 000 registos) e validação cruzada de 3 folds;
2. **TunedThresholdClassifierCV** — integra a otimização do threshold de decisão na validação cruzada com `scoring='f1'`, evitando o threshold 0.5 inadequado.


```python
X_gs_canc = X_train_cancelled.sample(frac=0.30, random_state=42).astype('float32')
y_gs_canc = y_train_cancelled.loc[X_gs_canc.index]
cv3 = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

print(f'Amostra GridSearch: {X_gs_canc.shape[0]:,} registos | '
      f'Positivos: {y_gs_canc.sum():,} ({y_gs_canc.mean():.2%})')
print(f'Modelo a otimizar: {melhor_nome_canc}')
print()

if melhor_nome_canc == 'XGBoost':
    base_gs_canc = XGBClassifier(
        scale_pos_weight=spw_cancelled, eval_metric='aucpr',
        random_state=42, n_jobs=1, verbosity=0)
    grid_canc = {
        'learning_rate': [0.05, 0.10],
        'max_depth':     [6, 8],
        'n_estimators':  [200],
        'subsample':     [0.8],
        'colsample_bytree': [0.8],
    }
else:
    base_gs_canc = HistGradientBoostingClassifier(
        class_weight='balanced', random_state=42)
    grid_canc = {
        'learning_rate':    [0.05, 0.10],
        'max_depth':        [6, 10],
        'max_iter':         [200],
        'min_samples_leaf': [30],
    }

gs_canc = GridSearchCV(
    base_gs_canc, grid_canc,
    scoring='average_precision', cv=cv3,
    n_jobs=1, refit=True, verbose=0
)
gs_canc.fit(X_gs_canc, y_gs_canc)

res_gs_canc = pd.DataFrame(gs_canc.cv_results_)[[
    'params', 'mean_test_score', 'rank_test_score'
]].sort_values('rank_test_score')
res_gs_canc['mean_test_score'] = res_gs_canc['mean_test_score'].round(4)
display(res_gs_canc.reset_index(drop=True).head(4))
print(f'\nMelhor AP (CV): {gs_canc.best_score_:.4f}')
print(f'Melhores parâmetros: {gs_canc.best_params_}')
```

```python
X_tuned_canc = X_train_cancelled.sample(frac=0.30, random_state=42).astype('float32')
y_tuned_canc = y_train_cancelled.loc[X_tuned_canc.index]

if melhor_nome_canc == 'XGBoost':
    est_canc = XGBClassifier(
        scale_pos_weight=spw_cancelled, eval_metric='aucpr',
        random_state=42, n_jobs=1, verbosity=0,
        **gs_canc.best_params_)
else:
    est_canc = HistGradientBoostingClassifier(
        class_weight='balanced', random_state=42,
        **gs_canc.best_params_)

modelo_final_cancelled = TunedThresholdClassifierCV(
    estimator=est_canc,
    scoring='f1',
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    thresholds=30,
    refit=True,
)
modelo_final_cancelled.fit(X_tuned_canc, y_tuned_canc)
threshold_cancelled = modelo_final_cancelled.best_threshold_
print(f'Threshold otimizado por CV (F1): {threshold_cancelled:.4f}')
print(f'F1 de CV no threshold otimizado: {modelo_final_cancelled.best_score_:.4f}')
```

O GridSearch seleciona `learning_rate=0.1` e `max_depth=10` com Average Precision de CV de **0.075** — margem consistente face ao candidato base (AP=0.096 no teste completo). O TunedThresholdClassifierCV estabiliza o threshold em **0.806** com F1 de CV de **0.138**, próximo dos 0.151 obtidos no conjunto de teste — indicando boa generalização e ausência de overfitting no processo de otimização.

## 3. Modelação — Previsão de Atrasos (`is_delayed`)

Esta secção modela `is_delayed`: atraso de origem meteorológica ou por aeronave anterior superior a 15 minutos, definido apenas para voos não cancelados.

Toda a nomenclatura de variáveis utiliza o sufixo `_delayed` para separação explícita.


### 3.1 Preparação dos Dados

```python
# Apenas voos não cancelados — evita leakage (is_delayed=0 para todos os cancelados)
df_delayed = df[df['cancelled'] == 0].copy()

X_delayed = df_delayed[feature_cols].copy()
y_delayed = df_delayed['is_delayed'].copy()

X_train_delayed, X_test_delayed, y_train_delayed, y_test_delayed = train_test_split(
    X_delayed, y_delayed,
    test_size=0.20, random_state=42, stratify=y_delayed
)

spw_delayed = float((y_train_delayed == 0).sum() / (y_train_delayed == 1).sum())

divisao_del = pd.DataFrame({
    'Conjunto': ['Treino', 'Teste'],
    'Registos': [X_train_delayed.shape[0], X_test_delayed.shape[0]],
    'Positivos': [int(y_train_delayed.sum()), int(y_test_delayed.sum())],
    'Prevalencia': [
        f'{y_train_delayed.mean():.3%}',
        f'{y_test_delayed.mean():.3%}',
    ],
})
display(divisao_del.set_index('Conjunto'))
print(f'\nscale_pos_weight para XGBoost: {spw_delayed:.1f}')
```

O conjunto de trabalho exclui os 15 905 cancelamentos, ficando com 1 025 246 voos. Destes, 88 690 são atrasos positivos (8.651%) — prevalência cerca de 5.5 vezes superior à de `cancelled`. O `scale_pos_weight` de **10.6** é 6 vezes inferior ao de `cancelled` (64.5), refletindo o menor desequilíbrio e a maior disponibilidade de exemplos positivos para aprendizado.


### 3.2 Modelo Baseline

```python
baseline_delayed = LogisticRegression(
    class_weight='balanced', max_iter=1000, random_state=42, n_jobs=-1
)
baseline_delayed.fit(X_train_delayed.astype('float32'), y_train_delayed)

y_proba_bl_del = baseline_delayed.predict_proba(
    X_test_delayed.astype('float32'))[:, 1]
thr_bl_del = threshold_otimo_f1(y_test_delayed, y_proba_bl_del)
y_pred_bl_del = (y_proba_bl_del >= thr_bl_del).astype(int)

print(f'Threshold otimizado (F1): {thr_bl_del:.4f}')
print()
display(relatorio_df(y_test_delayed, y_pred_bl_del,
                     ['Não Atrasado', 'Atrasado']))
```

Com threshold de **0.515** — próximo do 0.5 convencional — a Regressão Logística atinge Recall=0.565 e Precision=0.123 (F1=0.202). O threshold mais baixo face ao de `cancelled` (0.733) confirma que o sinal de atraso é mais forte e mais detetável por um modelo linear. Ainda assim, a Precision de 12.3% traduz-se em aproximadamente 7 alarmes falsos por cada atraso corretamente identificado.


### 3.3 Comparação de Modelos Candidatos

```python
candidatos_delayed = {
    'Logistic Regression': (baseline_delayed, y_proba_bl_del),
    'HistGradient Boosting': (
        HistGradientBoostingClassifier(
            class_weight='balanced', learning_rate=0.08,
            max_depth=8, max_iter=200, min_samples_leaf=30,
            random_state=42),
        None),
    'XGBoost': (
        XGBClassifier(
            scale_pos_weight=spw_delayed, learning_rate=0.08,
            max_depth=6, n_estimators=200, subsample=0.8,
            colsample_bytree=0.8, eval_metric='aucpr',
            random_state=42, n_jobs=1, verbosity=0),
        None),
}

resultados_del     = []
modelos_del        = {}
probabilidades_del = {}

for nome, (modelo, y_pr_cache) in candidatos_delayed.items():
    if y_pr_cache is not None:
        y_pr = y_pr_cache
    else:
        modelo.fit(X_train_delayed.astype('float32'), y_train_delayed)
        y_pr = modelo.predict_proba(X_test_delayed.astype('float32'))[:, 1]
    modelos_del[nome]        = modelo
    probabilidades_del[nome] = y_pr
    thr  = threshold_otimo_f1(y_test_delayed, y_pr)
    y_pd = (y_pr >= thr).astype(int)
    res  = {**avaliar(y_test_delayed, y_pd, y_pr, nome), 'Threshold': round(thr, 4)}
    resultados_del.append(res)

df_candidatos_del = (
    pd.DataFrame(resultados_del)
    .sort_values('Avg Precision', ascending=False)
    .reset_index(drop=True)
)
display(df_candidatos_del.set_index('Modelo').round(4))

melhor_nome_del = df_candidatos_del.iloc[0]['Modelo']
print(f'\nMelhor candidato (por Avg Precision): {melhor_nome_del}')
```

não precisa de ser tão conservador e consegue classificar atrasos com probabilidades mais baixas.


### 3.4 Otimização do Melhor Modelo

```python
X_gs_del = X_train_delayed.sample(frac=0.30, random_state=42).astype('float32')
y_gs_del = y_train_delayed.loc[X_gs_del.index]

print(f'Amostra GridSearch: {X_gs_del.shape[0]:,} registos | '
      f'Positivos: {y_gs_del.sum():,} ({y_gs_del.mean():.2%})')
print(f'Modelo a otimizar: {melhor_nome_del}')
print()

if melhor_nome_del == 'XGBoost':
    base_gs_del = XGBClassifier(
        scale_pos_weight=spw_delayed, eval_metric='aucpr',
        random_state=42, n_jobs=1, verbosity=0)
    grid_del = {
        'learning_rate': [0.05, 0.10],
        'max_depth':     [6, 8],
        'n_estimators':  [200],
        'subsample':     [0.8],
        'colsample_bytree': [0.8],
    }
else:
    base_gs_del = HistGradientBoostingClassifier(
        class_weight='balanced', random_state=42)
    grid_del = {
        'learning_rate':    [0.05, 0.10],
        'max_depth':        [6, 10],
        'max_iter':         [200],
        'min_samples_leaf': [30],
    }

gs_del = GridSearchCV(
    base_gs_del, grid_del,
    scoring='average_precision',
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    n_jobs=1, refit=True, verbose=0
)
gs_del.fit(X_gs_del, y_gs_del)

res_gs_del = pd.DataFrame(gs_del.cv_results_)[
    ['params', 'mean_test_score', 'rank_test_score']
].sort_values('rank_test_score').head(4).reset_index(drop=True)
res_gs_del['Hiperparâmetros'] = res_gs_del['params'].apply(
    lambda d: ', '.join(f'{k}={v}' for k, v in d.items())
)
res_gs_del = res_gs_del.rename(columns={'mean_test_score': 'AP médio (CV)'})[
    ['Hiperparâmetros', 'AP médio (CV)']
]
res_gs_del['AP médio (CV)'] = res_gs_del['AP médio (CV)'].round(4)
display(res_gs_del)
print(f'Melhor AP (CV): {gs_del.best_score_:.4f}')
print(f'Melhores parâmetros: {gs_del.best_params_}')
```

```python
X_tuned_del = X_train_delayed.sample(frac=0.30, random_state=42).astype('float32')
y_tuned_del = y_train_delayed.loc[X_tuned_del.index]

if melhor_nome_del == 'XGBoost':
    est_del = XGBClassifier(
        scale_pos_weight=spw_delayed, eval_metric='aucpr',
        random_state=42, n_jobs=1, verbosity=0,
        **gs_del.best_params_)
else:
    est_del = HistGradientBoostingClassifier(
        class_weight='balanced', random_state=42,
        **gs_del.best_params_)

modelo_final_delayed = TunedThresholdClassifierCV(
    estimator=est_del,
    scoring='f1',
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    thresholds=30,
    refit=True,
)
modelo_final_delayed.fit(X_tuned_del, y_tuned_del)
threshold_delayed = modelo_final_delayed.best_threshold_
print(f'Threshold otimizado por CV (F1): {threshold_delayed:.4f}')
print(f'F1 de CV no threshold otimizado: {modelo_final_delayed.best_score_:.4f}')
```

Curiosamente, o GridSearch de `is_delayed` seleciona os mesmos hiperparâmetros que `cancelled` (`learning_rate=0.1`, `max_depth=10`), mas com AP de CV de **0.211** — 2.8× superior ao de `cancelled` (0.075), confirmando a maior previsibilidade estrutural dos atrasos.


### 3.5 Avaliação Final — is_delayed

### 4.2 Explicabilidade dos Modelos com SHAP

**SHAP (SHapley Additive exPlanations)** é uma abordagem baseada em teoria dos jogos que fornece explicações locais e globais para previsões de modelos de machine learning:

- **Valores SHAP**: Quantificam a contribuição de cada feature para mover a predição do valor base (média global) até a predição específica
- **Summary Plot**: Visualiza a importância agregada e o padrão de cada feature
- **Dependence Plot**: Mostra como mudanças numa feature específica afetam as previsões
- **Interpretabilidade**: Permite responder "por que o modelo previu X para este exemplo?"

> **Objetivo desta secção:** responder à pergunta *"o que leva o modelo a prever que um voo vai ser cancelado ou atrasar?"*  
> Cada gráfico vem acompanhado de uma interpretação simples.

```python
shap.initjs()

def amostra_estratificada(X, y, n_total=5000, frac_pos=0.20, seed=42):
    """Amostra com proporção mínima garantida de positivos."""
    rng = np.random.default_rng(seed)
    idx_pos = y[y == 1].index
    idx_neg = y[y == 0].index
    n_pos = min(len(idx_pos), int(n_total * frac_pos))
    n_neg = n_total - n_pos
    s_pos = rng.choice(idx_pos, size=n_pos, replace=False)
    s_neg = rng.choice(idx_neg, size=min(len(idx_neg), n_neg), replace=False)
    idx   = np.concatenate([s_pos, s_neg])
    rng.shuffle(idx)
    return X.loc[idx], y.loc[idx]

print('Setup concluído.')
```

### Gráfico 1 — Quais fatores influenciam mais o modelo?

**Como ler:** cada ponto é um voo. A posição horizontal mostra se aquele fator *aumentou* ou *reduziu* a probabilidade de cancelamento. A cor mostra o valor da variável — vermelho significa valor alto, azul significa valor baixo.


```python
fig, ax = plt.subplots(figsize=(12, 7))

# Amostra estratificada para SHAP
X_shap_canc, _ = amostra_estratificada(
    X_train_cancelled.astype("float32"),
    y_train_cancelled,
    n_total=5000,
    frac_pos=0.20,
    seed=42,
)

# SHAP values do melhor modelo para cancelled
explainer_canc = shap.Explainer(gs_canc.best_estimator_, X_shap_canc)
sv_canc = explainer_canc(X_shap_canc)

shap.summary_plot(
    sv_canc,
    X_shap_canc,
    max_display=10,
    plot_type="dot",
    show=False,
    plot_size=None,
)

ax = plt.gca()
ax.set_title(
    "O que leva um voo a ser CANCELADO\nTop 10 fatores mais influentes (2024)",
    fontsize=15, fontweight="bold", pad=18
)
ax.set_xlabel(
    "← Reduz a probabilidade de cancelamento  |  Aumenta a probabilidade de cancelamento →",
    fontsize=11, labelpad=12
)
ax.axvline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.4)

# Legenda simples
patch_alto = mpatches.Patch(color="#d73027", label="Valor alto da variável")
patch_baixo = mpatches.Patch(color="#4575b4", label="Valor baixo da variável")
ax.legend(handles=[patch_alto, patch_baixo], loc="lower right", fontsize=9)

plt.tight_layout()
plt.savefig("shap_beeswarm_cancelled.png", dpi=150, bbox_inches="tight")
plt.show()
print("Gráfico guardado: shap_beeswarm_cancelled.png")
```

**Interpretação**

| Fator | O que significa na prática |
|---|---|
| `month` — **pontos vermelhos à esquerda** | Meses de valor alto (meio/final do ano) *reduzem* o risco de cancelamento; meses iniciais (pontos azuis à direita, ex: janeiro/fevereiro, pico do inverno) *aumentam-no* substancialmente. |
| `day_of_month` — grande dispersão | É um fator com enorme volatilidade. O dia específico tem um impacto extremo — alguns dias protegem contra cancelamentos, enquanto outros disparam o risco. |
| `distance` — **pontos vermelhos à direita** | Voos mais longos (vermelho) têm uma maior probabilidade de serem cancelados em comparação com voos mais curtos. |
| Origens (`origin_ATL`, `DFW`, `CLT`, etc.) | Partir destes grandes hubs *reduz* o risco de cancelamento (pontos vermelhos todos posicionados à esquerda). Estes aeroportos protegem o voo. |

> 💡 **Ideia chave:** O modelo está a aprender que *quando* e *de onde* o voo parte é muito mais determinante do que para onde vai. A forte influência temporal (Mês, Dia do Mês e Dia da Semana são o Top 3) confirma que a data atua como o principal *proxy* do clima para o algoritmo. Curiosamente, embora grandes hubs (como DFW e CLT) possam gerar atrasos, eles funcionam como "escudos" contra cancelamentos, provavelmente porque têm mais tripulações, aviões de reserva e equipas de manutenção para salvar o voo em último recurso.

### Os 3 principais insights deste projeto

**1. O tempo domina tudo** — a data (mês, dia) é o fator mais importante. O modelo aprendeu que Janeiro = mais cancelamentos, porque Janeiro = mais neve. Sem dados meteorológicos reais, a data é o melhor proxy disponível.

**2. Cancelamentos e atrasos têm causas distintas** — os aeroportos que mais contribuem para cancelamentos (Buffalo, Midway) são diferentes dos que mais contribuem para atrasos (Chicago O'Hare, Dallas). Os primeiros têm clima severo; os segundos têm congestionamento operacional.

**3. O modelo sabe o que não sabe** — quando um voo é cancelado e o modelo não previu (Falso Negativo). Isso significa que o cancelamento foi por uma razão imprevisível — não é falha do modelo, é o limite do que os dados históricos conseguem capturar.

### 4.4 Reflexão Crítica

#### Desempenho obtido

Os resultados são o reflexo direto da estrutura dos dados. Para `cancelled`, o modelo consegue
ordenar os voos por risco (ROC-AUC=0.854) mas na prática deteta apenas 845 dos 3 181 cancelamentos,
gerando 8.5 falsos alarmes por cada deteção. Para `is_delayed` os resultados são substancialmente
melhores — 7 254 atrasos detetados com um rácio de 3.6 falsos alarmes por deteção. Esta diferença
não resulta de algoritmos diferentes nem de maior esforço de otimização: resulta exclusivamente
do facto de atrasos serem um evento muito mais frequente do que cancelamentos.

#### Limitações identificadas

**1. Desequilíbrio em `cancelled`:** a raridade extrema de cancelamentos cria um "teto" estrutural
para o F1-score, independentemente do algoritmo ou da otimização aplicada.

**2. Definição conservadora de `is_delayed`:** o target inclui apenas atrasos meteorológicos e por
aeronave anterior — uma parte significativa dos atrasos reais fica de fora, o que subestima a
prevalência real e limita o que o modelo pode aprender.

**3. Features exclusivamente pré-partida:** o modelo conhece a rota, o aeroporto, a data e a
distância — mas não sabe nada sobre o que está a acontecer no dia: condições meteorológicas,
tráfego aéreo ou o historial da aeronave nesse dia específico. São precisamente esses dados
que teriam maior poder preditivo.

**4. Janela temporal de apenas dois meses:** o dataset cobre janeiro e fevereiro de 2024, dois
meses de inverno norte-americano. Padrões de outras estações do ano não estão representados.

#### Reflexão metodológica

O contributo mais relevante deste trabalho é metodológico: pipeline sem leakage entre targets,
métricas adaptadas ao desequilíbrio, threshold otimizado em validação cruzada e interpretação
honesta dos resultados. Métricas baixas num problema estruturalmente limitado pelos dados
disponíveis não representam falha — representam transparência. Um resultado inflacionado
seria metodologicamente menos correto do que os valores reais obtidos.
